```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import tplquad
from scipy.special import spherical_jn
```

### the complete integral

$$U_D(R) = \int^\infty _0 \int^\infty _0 \int^\infty _0 k^2 j_0(kR) v_D(k) F(\rho) g(E)   r_1^2 j_0(kr_1) \rho_p(r_1)  r_2^2 j_0(kr_2) \rho_t(r_2) dk dr_1 dr_2$$

```python
def integrate(k, r2, r1,
              c1=11061.625,
              c2=2537.5,
              c=1,
              mu1=4,
              mu2=2.5,
              gamma=1,
              n=1,
              g=1,
              E=1,
              alpha=1):

    # Density functions
    def density_p(r):
        return np.exp(-alpha * r**2)

    def density_t(r):
        return np.exp(-alpha * r**2)

    # Total density
    rho = density_p(r1) + density_t(r2)

    # Fourier transform of M3Y interaction
    vD = (
        (4*np.pi*c1/mu1) / (mu1**2 + k**2)
        -
        (4*np.pi*c2/mu2) / (mu2**2 + k**2)
    )

    # Density-dependent factor
    F = c * (1 - gamma * rho**n)

    return (
        k**2
        * spherical_jn(0, 1)       # R = 1
        * vD
        * F
        * g
        * r1**2
        * r2**2
        * spherical_jn(0, k*r1)
        * spherical_jn(0, k*r2)
        * density_p(r1)
        * density_t(r2)
    )
```

```python
result, error = tplquad(
    integrate,

    # r1: outer integral
    0, 5,

    # r2: middle integral
    lambda r1: 0,
    lambda r1: 5,

    # k: inner integral
    lambda r1, r2: 0,
    lambda r1, r2: 8
)
```

```python
print("Result =", result)
print("Error  =", error)
```
