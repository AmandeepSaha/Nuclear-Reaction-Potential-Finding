```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import tplquad
from scipy.special import spherical_jn
```

# Double Folding model

$$ {\underline{\fbox{ $U_D = \int d^3\vec{r_1} \int d^3\vec{r_2} \rho_T(\vec{r_2}) v_d(\rho, E, s) \rho_p(\vec{r_1})$}}}$$

### $ U_D(R) $ Direct Part of $ \alpha + {}^{12}C $

$$U_D(R) = \int^\infty _0 \int^\infty _0 \int^\infty _0 \underbrace{k^2 j_0(kR) v_D(k) F(\rho) g(E)}_{I_{V_D}} \:\: \underbrace{ r_1^2 j_0(kr_1) \rho_p(r_1) }_{I_p} \:\: \underbrace{ r_2^2 j_0(kr_2) \rho_t(r_2)}_{I_T} dk dr_1 dr_2$$

### Potenial

$$V_D(s,\rho, E) = \bigg[11061.625 \frac{e^{-4s}}{4s} - 2537.5 \frac{e^{-2.5s}}{2.5s}\bigg]F(\rho)g(E) \\ \\ {\sf{where,}} \\ F(\rho) = c\bigg(1-\gamma \rho^n \bigg) \\ g(E) = \bigg(1-0.003 E_{AP} \bigg) \\ \sf{For, \: BDM3Y2, \: n = 2}$$

```python
def integrate(k, r2, r1,
              R,
              c1=11061.625,
              c2=2537.5,
              c=1.0664,
              mu1=4,
              mu2=2.5,
              gamma=6.0296,
              n=2):

    # Density functions
    def density_p(r):
        return 0.4229*np.exp(-0.7024 * r**2)

    def density_t(r, rho_0=0.207, a=0.425, R_0 = 2.1545):
        return rho_0/(1+np.exp((r-R_0)/a))
    
    # g(E)
    def g(E_lab, A=12):
        return (1-0.003*E_lab/A)

    # Total density
    rho = density_p(r1) + density_t(r2)

    # Fourier transform of M3Y interaction
    vD = (
        (4*np.pi*c1/mu1) / (mu1**2 + k**2)
        -
        (4*np.pi*c2/mu2) / (mu2**2 + k**2)
    )

    # Density-dependent factor
    F = c * (1 - gamma * rho**n)

    return 8*(
        k**2
        * spherical_jn(0, k*R)
        * vD
        * F
        * g(E_lab=141.7)
        * r1**2
        * r2**2
        * spherical_jn(0, k*r1)
        * spherical_jn(0, k*r2)
        * density_p(r1)
        * density_t(r2)
    )
```

```python
R_vals = np.array([0.5*i for i in range(21)])
```

```python
datas = []
errors = []
```

```python
for i in range(0,21):

    result, error = tplquad(
        integrate,

        # r1: outer integral
        0, 5,

        # r2: middle integral
        lambda r1: 0,
        lambda r1: 8,

        # k: inner integral
        lambda r1, r2: 0,
        lambda r1, r2: 4,

        args=(R_vals[i],)
    )
    
    datas.append(result)
    errors.append(error)
```
